```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
# Import data
data = pd.read_csv("DailyClimateDelhiTrain.csv")
print(data.head())
```

```python
# Descriptive Statistics of Data
print(data.describe())
```

```python
# info about columns in dataset.
print(data.info())
```

```python
# lets have a look at mean temperature
figure = px.line(data, x="date",
                 y="meantemp",
                 title="Mean Temperature in Delhi Over the Years")
figure.show()
```

```python
# Humidity in Delhi over the years
figure = px.line(data, x="date",
                 y="humidity",
                 title="Humidty in Delhi Over the Years")
figure.show()
```

```python
# Wind Speed
figure = px.line(data, x="date",
                 y="wind_speed",
                 title="Wind Speed in Delhi Over Period")
figure.show()
```

```python
# A look at relation between temperature and humidity.
figure = px.scatter(data_frame = data, 
                    x = "humidity",
                    y="meantemp", size="meantemp",
                    trendline="ols",
                    title="Connection between Temperature and Humidity")
figure.show()
```

```python
# Analyzing Temperature Change.
data["date"] = pd.to_datetime(data["date"])
data["year"] = data["date"].dt.year
data["month"] = data["date"].dt.month
print(data.head())
```

```python
# Teperature over different periods.
plt.style.use("fivethirtyeight")
plt.figure(figsize=(15, 10))
plt.title("Temperature Change in Delhi Over different Periods")
sns.lineplot(data=data, x='month', y='meantemp', hue='year')
plt.show()
```
